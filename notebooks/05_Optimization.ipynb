```python
%sql
EXPLAIN FORMATTED
SELECT
    s.seller_id,
    s.seller_state,
    SUM(s.total_item_value) AS total_revenue
FROM olist_ecommerce.silver.sales_detail s
GROUP BY
    s.seller_id,
    s.seller_state;
```

```python
%sql
-- Verificando se o Spark está usando o Broadcast.

EXPLAIN FORMATTED

SELECT
    s.seller_id,
    s.seller_state,
    SUM(s.total_item_value) AS total_revenue

FROM olist_ecommerce.silver.sales_detail s

INNER JOIN olist_ecommerce.silver.sellers_clean v
    ON s.seller_id = v.seller_id

GROUP BY
    s.seller_id,
    s.seller_state;
```
